### **Задача №1. Наивный поиск подстроки в строке [1 балла]**

<div align="justify">

Даны строки p и t. Требуется найти все вхождения строки p в строку t в качестве подстроки.

* **Формат ввода / входного файла (input.txt)**. Первая строка входного файла содержит p, вторая - t. Строки состоят из букв латинского алфавита.
* **Ограничения на входные данные**. 1 ≤ |p|, |t| ≤ 10<sup>4</sup>.

* **Формат вывода / выходного файла (output.txt)**. В первой строке выведите число вхождений строки p в строку t. Во второй строке выведите в возрастающем порядке номера символов строки t, с которых начинаются вхождения p. Символы нумеруются с единицы.

* **Ограничение по времени**. 2 сек.
* **Ограничение по памяти**. 256 мб.

</div>

**Листинг кода:**

```python
def simple_substring_search(pattern, text):
    matches = []
    len_pattern = len(pattern)
    len_text = len(text)

    # проходим по строке text до символа,
    # от которого длина искомого паттерна будет доставать до конца строки
    for i in range(len_text - len_pattern + 1):
        # проверяем, совпадает ли подстрока со строкой p
        if text[i:i + len_pattern] == pattern:
            # сохраняем позицию (нумерация с 1)
            matches.append(i + 1)

    return len(matches), matches


@time_and_memory_decorator
def main():
    with open('task1/input.txt', 'r') as file_input, open('task1/output.txt', 'w') as file_output:
        p = file_input.readline().strip()
        t = file_input.readline().strip()

        result = simple_substring_search(p, t)

        file_output.write(str(result[0]) + '\n')
        file_output.write(' '.join(map(str, result[1])) + '\n')


main()
```

**Текстовое объяснение решения:**

<div align="justify">

**Алгоритм простыми словами:** прикладываем шаблон pattern к каждой возможной позиции в строке text и проверяем их на совпадение.

**Алгоритм:**
1. проходим циклом по строке text от индекса 0 до len(text) - len(pattern), то есть до символа, от которого длина искомого паттерна будет доставать до конца строки
2. на каждой итерации берем срез - подстроку текста text такой же длины, как pattern
3. если подстрока совпадает с pattern, сохраняем текущий индекс + 1 (по условию нумерация индексов с 1)
4. выводим количество найденных вхождений и сами индексы

### **Задача №3. Паттерн в тексте [1 балла]**

<div align="justify">

В этой задаче ваша цель - реализовать алгоритм Рабина-Карпа для поиска заданного шаблона (паттерна) в заданном тексте.
* **Формат ввода / входного файла (input.txt)**. На входе две строки: паттерн P и текст T. Требуется найти все вхождения строки P в строку T в качестве подстроки.
* **Ограничения на входные данные**. 1 ≤ |P|, |T| ≤ 10<sup>6</sup>. Паттерн и текст содержат только латинские буквы.
* **Формат вывода / выходного файла (output.txt)**. В первой строке выведите число вхождений строки P в строку T. Во второй строке выведите в возрастающем порядке номера символов строки T, с которых начинаются вхождения P. Символы нумеруются с единицы.
* **Ограничение по времени**. 2 сек.
* **Ограничение по памяти**. 256 мб.

</div>

**Листинг кода:**

```python
def rabin_karp(pattern, text):
    m, n = len(pattern), len(text)
    if m > n: return []

    base = 53  # мощность алфавита = простое число, чуть больше размера алфавита (можно также base = random.randint(1, mod - 1))
    mod = 10 ** 9 + 7  # модуль = большое простое число, чтобы хэш не переполнялся и было меньше коллизий

    matches = []

    # предварительно вычисляем base^(m-1) % mod для удаления первого символа
    base_pow_m = 1
    for _ in range(m - 1):
        base_pow_m = (base_pow_m * base) % mod

    # хеши шаблона и первой подстроки текста
    pattern_hash = 0
    window_hash = 0

    # вычисляем начальные хеши за O(M)
    for i in range(m):
        pattern_hash = (pattern_hash * base + ord(pattern[i])) % mod
        window_hash = (window_hash * base + ord(text[i])) % mod

    # проходим скокользящим окном по тексту за O(N)
    for i in range(n - m + 1):
        # если хеши совпали, проверяем равенство строк для защиты от коллизий
        if pattern_hash == window_hash:
            if text[i:i + m] == pattern:
                # сохраняем позицию (нумерация с 1)
                matches.append(i + 1)

        # пересчитываем хеш для следующего окна за O(1)
        if i < n - m:
            window_hash = (window_hash - ord(text[i]) * base_pow_m) % mod  # удаляем старый первый символ подстроки
            window_hash = (window_hash * base +
                           ord(text[i + m])) % mod  # делаем сдвиг разрядов подстроки и добавляем новый символ к окну

    return len(matches), matches


@time_and_memory_decorator
def main():
    with open('task3/input.txt', 'r') as file_input, open('task3/output.txt', 'w') as file_output:
        p = file_input.readline().strip()
        t = file_input.readline().strip()

        result = rabin_karp(p, t)

        file_output.write(str(result[0]) + '\n')
        file_output.write(' '.join(map(str, result[1])) + '\n')


main()
```

**Текстовое объяснение решения:**

<div align="justify">

**Суть алгоритма Рабина-Карпа - вместо посимвольного сравнения, сравниваем и работаем с хешами строк:**
1. посчитать хеш шаблона P
2. посчитать хеши всех подстрок текста T длиной ∣P∣
3. сдвигать окно и обновлять хеш
    * для ускорения используем полиномиальный хеш и "скользящее окно": хеш следующей подстроки вычисляется на основе предыдущего за O(1).
4. выполняем проверку:
    * если хеши не равны, строки точно разные
    * если хеши равны, выполняем прямую проверку совпадения подстрок (чтобы исключить коллизии)

**Скользящее окно:**

При сдвиге окна на один символ:
* вычитаем значение первого символа: (old_hash - old_char ⋅ base<sup>m-1</sup>)
* умножаем результат на base (сдвигаем степени)
* добавляем значение нового символа
* берем остаток по модулю p

Например, при переходе окна от 'abc' к 'bcd':
* убираем старый символ a
* умножаем на base
* добавляем новый символ (new_hash = (old_hash - old_char ⋅ base<sup>m-1</sup>) ⋅ base + new_char. Все берем по модулю p)

Замечание: при переходе от одного окна к другому нам не нужно пересчитывать его хеш с нуля за O(M). Мы можем выкинуть первый символ старого окна и добавить новый символ.

**Полиномиальный хеш:**

Строка 'abc' превращается в число: hash = (a ⋅ base<sup>m-1</sup> + b ⋅ base<sup>m-2</sup> + c ⋅ base<sup>0</sup>) mod p = ((a ⋅ base + b) ⋅ base + c) mod p.
* base - некоторое основание (обычно простое число, чуть больше размера алфавита)
* p - большой модуль (чтобы хэш не переполнялся и было меньше коллизий)

Алгоритм очень эффективен для поиска множества шаблонов одновременно (если они одной длины). Достаточно посчитать хеши всех шаблонов и проверять хеш окна текста на вхождение в этот набор. Однако алгоритм сильно зависит от хеш-функции и при неудачном выборе параметров можно получить много ложных срабатываний (прямых проверок после совпадения хэшей). Также требуется посимвольная проверка при совпадении хеша.

### **Задача №7. Наибольшая общая подстрока [2 балла]**

<div align="justify">

В задаче на наибольшую общую подстроку даются две строки s и t, и цель состоит в том, чтобы найти строку w максимальной длины, которая является подстрокой как s, так и t. Это естественная мера сходства между двумя строками. Задача имеет применения для сравнения и сжатия текстов, а также в биоинформатике. Эту проблему можно рассматривать как частный случай проблемы расстояния редактирования (Левенштейна), где разрешены только вставки и удаления. Следовательно, ее можно решить за время O(|s||t|) с помощью динамического программирования. Есть также весьма нетривиальные структуры данных для решения этой задачи за линейное время O(|s| + |t|). В этой
задаче ваша цель - использовать хеширование для решения почти за линейное время.

* **Формат ввода / входного файла (input.txt)**.Каждая строка входных данных содержит две строки s и t, состоящие из строчных латинских букв.
* **Ограничения на входные данные**. Суммарная длина всех s, а также суммарная длина всех s не превышает 100000.
* **Формат вывода / выходного файла (output.txt)**. Для каждой пары строк s<sub>i</sub> и t<sub>i</sub> найдите ее самую длинную общую подстроку и уточните ее параметры, выведя три целых числа: ее начальную позицию в s, ее начальную позицию в t (обе считаются с 0) и ее длину. Формально выведите целые числа 0 ≤ i < |s|, 0 ≤ j < |t| и l ≥ 0 такие, что и l максимально. (Как обычно, если таких троек с максимальным l много, выведите любую из них.)
* **Ограничение по времени**. 15 сек.
* **Ограничение по памяти**. 512 мб.

</div>

**Листинг кода:**

```python
import random

low, high = 10 ** 6, 10 ** 9  # нижняя и верхняя границы полиномиального хеша соответственно
base = random.randint(low, high)  # основание полиномиального хеша
mod1 = 10 ** 9 + 7  # модуль1 = большое простое число, чтобы число не переполнялось и было меньше коллизий
mod2 = 10 ** 9 + 9  # модуль2 = большое простое число, чтобы число не переполнялось и было меньше коллизий


def build_prefix_hashes_and_powers(string):
    """
    Вычисление префиксных хешей для подстрок разной длины и степеней основания

    Сложность: O(n)
    """
    n = len(string)  # длина строки

    # массив префиксных хешей hash[i] = хеш строки string[0:i], где i - длина префикса
    hash1 = [0] * (n + 1)  # префиксный хеш, построенный на mod1
    hash2 = [0] * (n + 1)  # префиксный хеш, построенный на mod2

    # массив степеней base power[i] = base^i % mod
    power1 = [1] * (n + 1)  # массив степеней, построенный на mod1
    power2 = [1] * (n + 1)  # массив степеней, построенный на mod2

    for i in range(n):
        x = ord(string[i])  # переводим символ в число

        # строим хеш как число: hash[i+1] = hash[i] * base + string[i]
        hash1[i + 1] = (hash1[i] * base + x) % mod1
        hash2[i + 1] = (hash2[i] * base + x) % mod2

        # считаем степени base
        power1[i + 1] = (power1[i] * base) % mod1
        power2[i + 1] = (power2[i] * base) % mod2

    return hash1, hash2, power1, power2


def get_hash(prefix_hashes, power, left, right, mod):
    """
    Получение хеша подстроки string[left:right], причем left и right включительно

    Сложность: O(1)
    """
    # hash(string[left:right]) = hash(0:right) - hash(0:left) * base^(right-left)
    return (prefix_hashes[right] - prefix_hashes[left] * power[right - left]) % mod


def check_common_substring(string1, string2, hash_string1, hash_string2, power_string1, power_string2,
                           estimated_length):
    """
    Проверка - есть ли общая подстрока длины estimated_length

    Сложность: O(n)
    """
    # если предполагаемая длина = 0, то всегда есть общая ПУСТАЯ строка
    if estimated_length == 0:
        return 0, 0

    # распаковка хешей строк
    hash1_string1, hash2_string1 = hash_string1
    hash1_string2, hash2_string2 = hash_string2
    # распаковка степеней строк
    power1_string1, power2_string1 = power_string1
    power1_string2, power2_string2 = power_string2

    dict_hash_string1 = {}  # словарь хешей вида - ((хеш1, хеш2): стартовая позиция в string1)

    # перебираем все подстроки строки string1 длины estimated_length
    for i in range(len(string1) - estimated_length + 1):
        # получаем два типа хешей для выделенной подстроки
        h1 = get_hash(hash1_string1, power1_string1, i, i + estimated_length, mod1)
        h2 = get_hash(hash2_string1, power2_string1, i, i + estimated_length, mod2)
        # запоминаем по ключу (h1, h2) индекс начала хеша в строке
        dict_hash_string1[(h1, h2)] = i

    # перебираем все подстроки строки string2 длины estimated_length
    for j in range(len(string2) - estimated_length + 1):
        # получаем два типа хешей для выделенной подстроки
        h1 = get_hash(hash1_string2, power1_string2, j, j + estimated_length, mod1)
        h2 = get_hash(hash2_string2, power2_string2, j, j + estimated_length, mod2)

        # если нашли совпадение хешей
        if (h1, h2) in dict_hash_string1:
            # возвращаем позицию хеша в string1 и позицию в string2
            return dict_hash_string1[(h1, h2)], j

    # если не встретили совпадающих хешей
    return -1, -1


def find_largest_common_substring(string1, string2):
    """
    Нахождение наибольшей общей подстроки с помощью бинарного поиска по ответу

    Сложность: O(log n)
    """
    # считаем префиксные хеши двух типов и степени оснований для обеих строк
    hash1_string1, hash2_string1, power1_string1, power2_string1 = build_prefix_hashes_and_powers(string1)
    hash1_string2, hash2_string2, power1_string2, power2_string2 = build_prefix_hashes_and_powers(string2)

    # длина общей подстроки может варьироваться от 0 до min(len(string1), len(string2))
    left, right = 0, min(len(string1), len(string2))

    result = (0, 0, 0)  # (i, j, длина наибольшей общей подстроки)

    # двочиный поиск по длине общей подстроки
    while left <= right:
        # пробуем длину mid
        mid = (left + right) // 2
        # проверяем существование общей подстроки длины mid
        i, j = check_common_substring(string1, string2, (hash1_string1, hash2_string1), (hash1_string2, hash2_string2),
                                      (power1_string1, power2_string1), (power1_string2, power2_string2), mid)

        # если нашли подстроку
        if i != -1:
            result = (i, j, mid)  # сохраняем текущий найденный максимум
            left = mid + 1  # пробуем найти общую подстроку большей длины (ищем максимум)
        else:
            right = mid - 1  # уменьшаем длину

    return result


@time_and_memory_decorator
def main():
    with open('task7/input.txt', 'r') as file_input, open('task7/output.txt', 'w') as file_output:
        while line := file_input.readline():
            s1, s2 = line.split()

            i, j, l = find_largest_common_substring(s1, s2)

            file_output.write(f'{i} {j} {l}' + '\n')


main()
```

**Текстовое объяснение решения:**

<div align="justify">

Даны строки: string1 и string2. Нужно найти максимальную длину length такую, что существует: string1[i : i+length] = string2[j : j+length].


**Алгоритм поиска наибольшей общей подстроки кратко:**
1. предобработка: считаем префиксные хеши и степени основания
2. запускаем бинарный поиск по длине length
3. для каждого length:
   * строим хеши всех подстрок string1
   * проверяем string2
4. если нашли совпадение хешей, то увеличиваем length
5. если нет, то уменьшаем length
6. возвращаем максимум


**Шаги алгоритма поиска наибольшей общей подстроки:**
1. делаем предобработку для использования скользящего хеша, то есть строим префиксные хеши для string1 и string2, а также сразу считаем степени основания:
    * hash[i] = hash(string[0:i])
    * power[i] = base^i

    Это позволяет быстро считать hash(string[left:right]) за O(1)
2. бинарный поиск по длине length (length ∈ [0, min(|string1|, |string2|)])

    Замечание: если существует общая подстрока длины length, то обязательно существуют и подстроки длины: length-1, length-2, ..., 0. Это дает монотонность, значит можно применить бинарный поиск по ответу.
    * на каждом шаге:
        * берем mid = (left + right) // 2 и проверяем есть ли общая подстрока длины mid
3. проверка длины length (функция check_common_substring) - есть ли подстрока длины length, которая встречается и в string1, и в string2
    * перебираем i = 0 ... (|string1| - length)
    * считаем hash(string1[i : i+length]) и сохраняем в хеш-таблицу:
        * key = (h1, h2): value = i, то есть подстрока длины length с хешем key начинается в позиции i в string1
4. проверка строки string2
    * перебираем j = 0 ... (|string2| - length)
    * считаем hash(string2[j : j+length])
    * проверяем: hash(string2[j:j+length]) ∈ hashes_s
        * если ДА, значит string1[i:i+length] = string2[j:j+length] (с высокой вероятностью благодаря использованию двух разных хеш функций), возвращаем (i, j)

        Замечание: используем двойное хеширование, так как один хеш может совпасть у разных строк (например hash('abc') = hash('xyz')). Используем два хеша (h1, h2), где mod1 = 10^9 + 7, а mod2 = 10^9 + 9, тогда если хеши двух подстрок совпали, то они почти точно равны.

        * если НЕТ, возвращаем (-1, -1)
5. после проверки:
    * если нашли length = mid, тогда пробуем найти общую подстроку большей длины: left = mid + 1
    * если не нашли, то length слишком большой, уменьшаем: right = mid - 1

    В процессе храним текущий результат - (i, j, length).

Формула получения хеша подстроки (убирания левой части подстроки): hash(left, right) = hash[right] - hash[left] * base^(right-left), где hash[right] = hash(0:right), hash[left] = hash(0:left). Асимптотика - O(1).

### **Задача №2. Карта [1 балла]**

<div align="justify">

В далеком 1744 году во время долгого плавания в руки капитана Александра Смоллетта попала древняя карта с указанием местонахождения сокровищ. Однако расшифровать ее содержание было не так уж и просто.

Команда Александра Смоллетта догадалась, что сокровища находятся на x шагов восточнее красного креста, однако определить значение числа она не смогла. По возвращению на материк Александр Смоллетт решил обратиться за помощью в расшифровке послания к знакомому мудрецу. Мудрец поведал, что данное послание таит за собой некоторое число. Для вычисления этого числа необходимо было удалить все пробелы между словами, а потом посчитать количество способов вычеркнуть все буквы кроме трех так, чтобы полученное слово из трех букв одинаково читалось слева направо и справа налево.

Александр Смоллетт догадывался, что число, зашифрованное в послании, и есть число x. Однако, вычислить это число у него не получилось. После смерти капитана карта была безнадежно утеряна до тех пор, пока не оказалась в ваших руках. Вы уже знаете все секреты, осталось только вычислить число x.

* **Формат ввода / входного файла (input.txt)**. В единственной строке входного файла дано послание, написанное на карте.
* **Ограничения на входные данные**. Длина послания не превышает 3 · 10<sup>5</sup>. Гарантируется, что послание может содержать только строчные буквы английского алфавита и пробелы. Также гарантируется, что послание не пусто. Послание не может начинаться с пробела или заканчиваться им.
* **Формат вывода / выходного файла (output.txt)**. Выведите одно число x - число способов вычеркнуть из послания все буквы кроме трех так, чтобы оставшееся слово одинаково читалось слева направо и справа налево.
* **Ограничение по времени**. 2 сек.
* **Ограничение по памяти**. 256 мб.

**Листинг кода:**

```python
def count_palindromes_len3(string):
    # удаляем все пробелы из строки, преобразуя ее в массив
    arr = [char for char in string if char != ' ']

    left = [0] * 26  # сколько букв char уже слева от текущей позиции
    right = [0] * 26  # сколько букв char еще осталось справа от текущей позиции

    # заполняем массив right - сначала все буквы справа
    for char in arr:
        # преобразуем символ в число и по нему сохраняем в right кол-во вхождений этого символа в строку
        right[ord(char) - ord('a')] += 1

    count = 0  # счетчик

    # идем по строке
    for char in arr:
        # текущая позиция
        cur_center = ord(char) - ord('a')

        # убираем текущий символ из right, так как он уже не справа, а на нашей текущей позиции
        right[cur_center] -= 1

        # считаем все возможные сочетания для каждой буквы слева и справа от текущей
        for c in range(26):
            count += left[c] * right[c]

        # добавляем текущий символ в left
        left[cur_center] += 1

    return count


@time_and_memory_decorator
def main():
    with open('task2/input.txt', 'r') as file_input, open('task2/output.txt', 'w') as file_output:
        s = file_input.readline()
        result = count_palindromes_len3(s)
        file_output.write(str(result))


main()
```

**Текстовое объяснение решения:**

<div align="justify">

**Алгоритм нахождения кол-ва палиндромов длины 3**:
1. удаляем из строки все пробелы
2. инициализируем два массива:
    * left[char] - сколько раз символ char встречается слева от текущей позиции
    * right[chat] - сколько раз символ char встречается справа от текущей позиции
3. изначально заполняем right количеством вхождений каждого символа в строку
4. последовательно рассматриваем каждый символ как середину палиндрома длины 3
5. для каждого символа char добавляем к count (число всех палиндромов длины 3) left[c] * right[c] (количество способов выбрать одинаковый символ слева и справа)
6. после обработки переносим текущий символ из right в left
